```python
import google.generativeai as genai

# Inicializar modelo
model = genai.GenerativeModel("gemini-3-flash-preview")

BASE_PROMPT = """
Actúa como asistente experto en organización personal adaptativa.

Organiza una semana equilibrada considerando:

HORARIOS FIJOS:
{horarios_fijos}

HORARIOS FLEXIBLES:
{horarios_flexibles}

HÁBITOS:
{habitos}

NIVEL DE ENERGÍA:
{energia}

INSTRUCCIONES:
- Prioriza tareas importantes.
- Evita sobrecarga diaria.
- Agrupa tareas similares.
- Mantén equilibrio entre trabajo, estudio y descanso.

Devuelve:
1. Plan semanal estructurado por día.
2. Justificación breve de la organización.
"""
```

```python
def generar_plan(horarios_fijos, horarios_flexibles, habitos, energia):

    prompt = BASE_PROMPT.format(
        horarios_fijos=horarios_fijos,
        horarios_flexibles=horarios_flexibles,
        habitos=habitos,
        energia=energia
    )

    response = model.generate_content(
        prompt,
        generation_config={
            "temperature": 0.3
        }
    )

    return response.text
```

```python
def replanificar(plan_original, cambio):

    prompt = f"""
Actúa como asistente organizativo adaptativo.

PLAN ORIGINAL:
{plan_original}

CAMBIO IMPREVISTO:
{cambio}

Reorganiza la semana manteniendo prioridades y evitando sobrecarga.
Devuelve solo el nuevo plan actualizado.
"""

    response = model.generate_content(
        prompt,
        generation_config={
            "temperature": 0.2
        }
    )

    return response.text
```

```python
def generar_prompt_imagen(plan_semanal):

    prompt = f"""
Eres un diseñador experto en visualizaciones de productividad.

A partir del siguiente plan semanal, crea un prompt para generar una imagen
de un calendario semanal visual.

PLAN SEMANAL:
{plan_semanal}

El prompt debe generar:

- Un calendario semanal tipo planner
- Columnas para: Monday, Tuesday, Wednesday, Thursday, Friday, Saturday, Sunday
- Filas con horarios del día
- Bloques de colores distintos para cada tipo de actividad:
  - Estudios (verde)
  - Edición / trabajo (naranja)
  - Ejercicio (morado)
  - Descanso / ocio (azul)

Estilo visual:
- Minimalista
- Limpio
- Tipo app de productividad
- Vista tipo Notion / Google Calendar
- Fondo claro

IMPORTANTE:
El prompt debe reflejar exactamente las actividades del plan.

Devuelve SOLO el prompt de imagen.
"""

    response = model.generate_content(
        prompt,
        generation_config={"temperature": 0.6}
    )

    return response.text
```

```python
# EJEMPLO DE USO

plan = generar_plan(
    horarios_fijos="Lunes a viernes 9-13 estudios",
    horarios_flexibles="Edición freelance",
    habitos="Ejercicio 3 veces por semana",
    energia="Mañana alta, tarde media, noche baja"
)

print(plan)

prompt_imagen = generar_prompt_imagen(plan)

print(prompt_imagen)
```
